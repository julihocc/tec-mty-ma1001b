```python
import numpy as np
import pandas as pd
from scipy import stats

rng = np.random.default_rng(1001)
n_trials = 20
p_success = 0.35
simulated_successes = rng.binomial(n=n_trials, p=p_success, size=5000)
counts = pd.Series(simulated_successes).value_counts(normalize=True).sort_index()
counts.head()
```

## Guided Practice


```python
x = np.arange(0, n_trials + 1)
binomial_table = pd.DataFrame({
    "successes": x,
    "theoretical_probability": stats.binom.pmf(x, n_trials, p_success),
    "empirical_probability": counts.reindex(x, fill_value=0).to_numpy(),
})
binomial_table.head(10)
```
