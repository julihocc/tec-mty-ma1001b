## Data Context

Kaggle House Prices, expected at data/raw/house-prices/train.csv.

If the expected real dataset is missing, this notebook uses a small simulated
dataset so the statistical workflow remains runnable. Replace the simulated
data with the real dataset before submitting graded work.


```python
from pathlib import Path
import numpy as np
import pandas as pd
import seaborn as sns

rng = np.random.default_rng(1001)
path = Path("data/raw/house-prices/train.csv")

if path.exists():
    homes = pd.read_csv(path)
    values = homes["SalePrice"].dropna()
else:
    print(f"Missing {path}. Download House Prices from https://www.kaggle.com/c/house-prices-advanced-regression-techniques")
    values = pd.Series(rng.lognormal(mean=12.0, sigma=0.35, size=1200), name="SalePrice")

values.describe()
```

## Guided Practice


```python
ax = sns.histplot(values, kde=True)
ax.set_title("Continuous outcome distribution")
ax.set_xlabel(values.name or "value")
```
