```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

rng = np.random.default_rng(1001)
sns.set_theme(style="whitegrid")

sample = pd.DataFrame({
    "team": np.repeat(["A", "B", "C"], 30),
    "hours_studied": rng.normal(7.5, 1.8, 90).clip(1),
    "quiz_score": rng.normal(78, 9, 90).clip(40, 100),
    "submitted": rng.choice([True, False], 90, p=[0.86, 0.14])
})

sample.head()
```

## Guided Practice


```python
summary = (
    sample
    .groupby("team")
    .agg(
        students=("quiz_score", "size"),
        mean_score=("quiz_score", "mean"),
        sd_score=("quiz_score", "std"),
        submission_rate=("submitted", "mean"),
    )
    .round(2)
)
summary
```
